```python
import pandas as pd

practice_df = pd.DataFrame({
    'player_name': ['Alice', 'Bob', 'Carol'],
    'match_id': [1001, 1002, 1003],
    'purchases': [
        [{'time': 0, 'item': 'sword', 'cost': 100}, {'time': 30, 'item': 'shield', 'cost': 150}],
        [{'time': 5, 'item': 'potion', 'cost': 20}],
        [{'time': 10, 'item': 'bow', 'cost': 200}, {'time': 45, 'item': 'arrow', 'cost': 10}, {'time': 60, 'item': 'quiver', 'cost': 50}],
    ]
})

practice_df
```

```python
exploded = practice_df.explode('purchases').reset_index(drop=True)
exploded['purchases']
```

```python
df = pd.json_normalize(exploded['purchases'])
```

```python
fixed = exploded.join(df)
fixed
```
