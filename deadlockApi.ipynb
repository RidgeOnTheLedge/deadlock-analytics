```python
import requests, json
import pandas as pd
pd.set_option("display.max_rows", 500)
pd.set_option("display.max_columns", 500)

url  = 'https://api.deadlock-api.com/v1/matches/metadata'


params = {
    "limit": 500,
    "order_by": "start_time",
    "order_direction": "desc",
    "match_mode": "ranked",
    "include_info": "true",
    "include_more_info": "true",
    "include_objectives": "true",
    "include_mid_boss": "true",
    "include_player_info": "true",
    "include_player_final_stats": "true",
    "include_player_stats": "true",
    "include_player_items": "true",
    "include_player_death_details": "true",
    "hero_ids": "77", # Only get hero ids for apollo
    "format": "json",
}
response = requests.get(url, params=params, timeout=30)
response.raise_for_status()
matches = response.json()

# print(len(matches))
# print(json.dumps(matches, indent=2)[:3000])
df = pd.DataFrame(matches)
df.head()
```

```python
url = 'https://api.deadlock-api.com/v1/assets/items'
response = requests.get(url, timeout=30)
response.raise_for_status()
items = response.json()

mapping = {}
for item in items:
    if item['type'] == 'upgrade' and item.get('shopable', False):
        mapping[item["id"]] = item["name"]

items_df["item_name"] = items_df["item_id"].map(mapping)
```

```python
url = "https://api.deadlock-api.com/v1/analytics/item-stats"

params = {
    "hero_id": 77,
    "match_mode": "ranked",
}
response = requests.get(url, timeout=30)
response.raise_for_status()
item_stats = response.json()

# print(len(matches))
# print(json.dumps(matches, indent=2)[:3000])
item_stats_df = pd.DataFrame(item_stats)
item_stats_df.head()


item_stats_df["item_name"] = item_stats_df["item_id"].map(mapping)
item_stats_df.head(1)
```

```python
wins_mean = item_stats_df["wins"].mean()
item_stats_df["abv_avr"] = item_stats_df["wins"] >= wins_mean
items_df["is_shop_item"] = items_df["item_name"].notna()
item_stats_df["is_shop_item"] = items_df["item_name"].notna()

df_items_shopable = (item_stats_df[(item_stats_df["is_shop_item"]) & (item_stats_df["abv_avr"])])

# sort
df_items_shopable.sort_values("win_rate", ascending=False, inplace=True)

y = df_items_shopable["win_rate"]
x = df_items_shopable["item_name"]
z = df_items_shopable["avg_buy_time_s"]


import plotly.express as px


df = px.data.iris()
fig = px.scatter(df, x=z, y=y, color=x, symbol=x)
fig.update_layout(xaxis_title="Avg Buy Time Seconds", yaxis_title="Win Rate")
fig.show()
```
